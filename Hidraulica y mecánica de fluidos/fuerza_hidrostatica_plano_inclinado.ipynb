```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.cm as cm


def h_centroide(theta, l, h_uno, rho_uno, rho_dos):

   #altura del centroide en la parte inferior de la superficie
   #Los dos líquidos generan una fuerza en la parte inferior de la superficie
   #se crea una nueva altura que relacione ambas densidades

  nueva_h_uno = (rho_uno/rho_dos)*h_uno

   #altura del centroide en la parte superior de la superficie inclinada
   #existen 2 situaciones, cuando la altura del liquido es mayor a la altura de la parte superior de superficice inclinada
   #cuando la altura del liquidio es menor a la altura de la parte superior de superficice inclinada

  hc_dos = np.sin(theta)*(l/4) + nueva_h_uno

  if h_uno >= (l/2)*np.sin(theta):
      hc_uno = h_uno - (l/4*np.sin(theta))
  else:
      hc_uno = (h_uno/2)

  return hc_uno, hc_dos

def fuerza_resultante(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta):

    #fuerza resultante superficie inferior
    area = b*l/2
    #fuerza resultante superficie superior
    #dos opciones

    fr_dos = rho_dos*g*(hc_dos)*area

    if h_uno >= (l/2)*np.sin(theta):
      fr_uno = rho_uno*g*hc_uno*area
    else:
      area = (h_uno/np.sin(theta))*b
      fr_uno = rho_uno*g*hc_uno*area

    return fr_uno, fr_dos


def ycp(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta):

  #distancia entre el centroide y punto donde  fuerza resultante interactua
  #distancia 2 nunca cambia
  #distancia 1 tiene dos opciones, cambia el area y momento de inercia dependiendo del ángulo
   area = b*l/2
   h = l/2
   ixx = (1/12)*b*(h**3)
   ycp_dos = (ixx*np.sin(theta))/(hc_dos*area)

   if h_uno >= (l/2)*np.sin(theta):
     ycp_uno = (ixx*np.sin(theta))/(hc_uno*area)
   else:
      h = (h_uno/np.sin(theta))
      area = b*h
      ixx = (1/12)*b*(h**3)
      ycp_uno = (ixx*np.sin(theta))/(hc_uno*area)

   return ycp_uno, ycp_dos

def coordenadas_fuerza_resultante(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta, fuerzas, distancias):

  #coordenadas individuales de fuerzas resultantes

  fr_2_point = (-(l/4+distancias[1])*np.cos(theta),-(l/4+distancias[1])*np.sin(theta))

  if h_uno >= (l/2)*np.sin(theta):
    fr_1_point = ((l/4-distancias[0])*np.cos(theta),(l/4-distancias[0])*np.sin(theta))
  else:
      h = (h_uno/2)/np.sin(theta)
      fr_1_point = ((h-distancias[0])*np.cos(theta),(h-distancias[0])*np.sin(theta))
  return fr_1_point, fr_2_point

def fuerza_resultante_total(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta, fuerzas, distancias):

  fr_uno_x = fuerzas[0]*np.cos(np.pi/2-theta) + fuerzas[1]*np.cos(np.pi/2-theta)
  fr_uno_y = fuerzas [0]*np.sin(np.pi/2-theta) + fuerzas[1]*np.sin(np.pi/2-theta)
  fuerza_resultante_total = np.sqrt(fr_uno_x**2 + fr_uno_y**2)

  return fuerza_resultante_total

def  coordenada_fuerza_resultante_total(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta, fuerzas, distancias, puntos,fuerza_resultante_total):
  fr_1 = fuerzas[0]
  fr_2 = fuerzas[1]
  fr_1x =abs(puntos[0][0])
  fr_1y =abs(puntos[0][1])
  fr_2x =abs(puntos[1][0])
  fr_2y =abs(puntos[1][1])

  #Sumatoria de momentos para encontrar posición de fuerza resultante
  distancia_fuerza_resultante_total= (-fr_1*((l/4)-distancias[0]) + fr_2*((l/4)+distancias[1]))/fuerza_resultante_total
  x =(-distancia_fuerza_resultante_total*np.cos(theta), -distancia_fuerza_resultante_total*np.sin(theta) )

  return x

def distribucion_presion(theta, l, h_uno,rho_uno,rho_dos):
  coordenadas_presion_x = []
  coordenadas_presion_y = []
  intensidades_presion = []
  modulo_x = []
  modulo_y = []
  inicio_x = []
  inicio_y = []

  factor = 50000
  incremento_uno = l/20
  incremento_dos = l/20
  dx = 0
  dx_l = 0
  nueva_h_uno = (rho_uno/rho_dos)*h_uno

  if h_uno >= (l/2)*np.sin(theta):
   while dx <= l / 2:
    y = (l / 2 - dx)
    altura_presion = (h_uno - y * np.sin(theta))
    presion = rho_uno * g * altura_presion
    coordenadas = (y*np.cos(theta), y*np.sin(theta))

    intensidad_presion = presion/factor
    x = y*np.cos(theta)
    y = y*np.sin(theta)
    coordenadas_presion_x.append(x)
    coordenadas_presion_y.append(y)

    inicio_x.append(x - (intensidad_presion*np.cos(np.pi/2-theta)))
    inicio_y.append(y + (intensidad_presion*np.sin(np.pi/2-theta)))
    modulo_x.append((intensidad_presion*np.cos(np.pi/2-theta)))
    modulo_y.append(-(intensidad_presion*np.sin(np.pi/2-theta)))

    intensidades_presion.append(intensidad_presion)

    dx += incremento_uno
  else:
    while dx <= h_uno:
     h_x= h_uno - dx
     presion = rho_uno * g * dx
     d = h_x/np.sin(theta)
     coordenadas = (d*np.cos(theta), h_x)

     intensidad_presion = presion/factor

     x = d*np.cos(theta)
     y = h_x

     coordenadas_presion_x.append(x)
     coordenadas_presion_y.append(y)

     inicio_x.append(x - ((presion/factor)*np.cos(np.pi/2-theta)))
     inicio_y.append(y + ((presion/factor)*np.sin(np.pi/2-theta)))
     modulo_x.append(((presion/factor)*np.cos(np.pi/2-theta)))
     modulo_y.append(-((presion/factor)*np.sin(np.pi/2-theta)))
     intensidades_presion.append(presion/factor)

     dx += incremento_dos

  while dx_l <= l/2:
   dx = dx_l*np.sin(theta)
   h = (nueva_h_uno + dx)
   presion = rho_dos * g * h
   coordenadas = (-dx_l*np.cos(theta), -dx_l*np.sin(theta))
   intensidad_presion = presion/factor

   x = -dx_l*np.cos(theta)
   y = -dx_l*np.sin(theta)

   coordenadas_presion_x.append(x)
   coordenadas_presion_y.append(y)

   inicio_x.append(x - ((intensidad_presion)*np.cos(np.pi/2-theta)))
   inicio_y.append(y + ((intensidad_presion)*np.sin(np.pi/2-theta)))

   modulo_x.append(((intensidad_presion)*np.cos(np.pi/2-theta)))
   modulo_y.append(-((intensidad_presion)*np.sin(np.pi/2-theta)))

   intensidades_presion.append(intensidad_presion)

   dx_l += incremento_uno

  return inicio_x, inicio_y, modulo_x, modulo_y


def grafico(puntos, b, coordenada_fuerza_resultante_total, l, theta, h_uno, h_dos, inicio_x,inicio_y,modulo_x,modulo_y, fuerza_resultante_total,i, angulo, rho_uno, rho_dos, fuerza_inicial):
    x = [puntos[0][0], puntos[1][0]]
    y = [puntos[0][1], puntos[1][1]]

    factor = (fuerza_resultante_total/fuerza_inicial)*(l/6)

    plt.figure(figsize=(10, 10))
    plt.axhspan(0, h_uno, facecolor='blue', alpha=0.3)
    plt.axhspan(0, -h_dos, facecolor='blue', alpha=0.6)
    plt.axhspan(-h_dos, -2*l/3, facecolor='brown', alpha=0.5)

    plt.plot([(l / 2) * np.cos(theta), -(l / 2) * np.cos(theta)], [(l / 2) * np.sin(theta), -(l / 2) * np.sin(theta)], label=f"Superficie, L = {l} [m], ancho = {b} [m]", color='black', linestyle='-')
    plt.scatter([coordenada_fuerza_resultante_total[0]], [coordenada_fuerza_resultante_total[1]], label=f"Fuerza resultante = {np.trunc(fuerza_resultante_total)} [N]", color='yellow')
    plt.scatter(x, y, label='Posición resultantes individuales', color='orange')
    plt.scatter([0], [0], label='Centroide superficie', color='red')
    plt.quiver(inicio_x, inicio_y, modulo_x, modulo_y, scale_units ='xy', scale = 1, label='Distribución de presiones')

    plt.arrow(coordenada_fuerza_resultante_total[0] - factor*np.cos(np.pi/2-theta), coordenada_fuerza_resultante_total[1] + factor*np.sin(np.pi/2-theta), factor*np.cos(np.pi/2-theta) ,-factor*np.sin(np.pi/2-theta), width=l*0.01, length_includes_head=True,color="yellow")

    plt.xlabel('')
    plt.ylabel('Profundidad [Metros]')
    plt.title('Fuerza resultante y distribución de presiones sobre superficie inclinada')

    plt.text(l/2, -l/3, f"{np.trunc(angulo)}°", color='white')
    plt.text(3*l/5, l/10, r"$\rho1$", color='white')
    plt.text(3*l/5, -l/10, r"$\rho2$", color='white')

    plt.legend()
    plt.grid(False)
    plt.xlim(-2*l/3, 2*l/3)
    plt.ylim(-2*l/3, 2*l/3)

    plt.savefig("C:\\Users\\chris\\Downloads\\Codigo" + "/file%02d.png" % i)

#Datos iniciales:
g = 9.81
theta = 0
angulo = 0
l = 20
h_uno = 5
h_dos = 25
rho_uno = 1000
rho_dos = 3500
b = 1
i = 0

#Programa principal
#900 imágenes para video de 30 imagenes por segundo por un total de 30 segundos

while h_dos < l/2 or h_uno > l/2:
    print("No se cumplen las condiciones iniciales")
    h_uno = int(input("h_uno:"))
    h_dos = int(input("h_dos:"))

fuerza_inicial = rho_uno*g*h_uno*l*b

while angulo <= 90:
  results = h_centroide(theta, l, h_uno, rho_uno, rho_dos)
  hc_uno = results[0]
  hc_dos = results[1]
  fuerzas = fuerza_resultante(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta)
  distancias = ycp(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta)
  puntos = coordenadas_fuerza_resultante(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta, fuerzas, distancias)
  x = fuerza_resultante_total(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta, fuerzas, distancias)
  y = coordenada_fuerza_resultante_total(l, b, rho_uno, rho_dos, hc_uno, hc_dos, h_uno, theta, fuerzas, distancias, puntos,x)
  datos_presion = distribucion_presion(theta, l, h_uno,rho_uno,rho_dos)
  inicio_x = datos_presion[0]
  inicio_y = datos_presion[1]
  modulo_x = datos_presion[2]
  modulo_y = datos_presion[3]
  grafico(puntos, b, y, l, theta, h_uno, h_dos, inicio_x,inicio_y,modulo_x,modulo_y, x, i, angulo, rho_uno, rho_dos, fuerza_inicial)
  theta += np.pi/1800
  angulo += 1/10
  i += 1


```
